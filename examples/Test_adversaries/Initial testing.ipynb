```python
def get_max_rec_errors(rec_error_dict):
    max_rec_errors = []
    for veh_id in rec_error_dict:
        rec_errors = rec_error_dict[veh_id][1]
        max_rec_errors.append(np.max(rec_errors))
        
    return max_rec_errors

all_max_rec_errors_without_attack = []
for rec_error_dict in rec_errors_without_attack:
    all_max_rec_errors_without_attack.append(get_max_rec_errors(rec_error_dict))
    
all_max_rec_errors_with_attack = []
for rec_error_dict in rec_errors_with_attack:
    all_max_rec_errors_with_attack.append(get_max_rec_errors(rec_error_dict))
    
```

# Get the AUC from a set of reconstruction errors:

```python
from sklearn import metrics

from sklearn.metrics import f1_score

def get_true_labels(veh_ids):
    veh_labels = []
    for veh_id in veh_ids:
        if(('adv' in veh_id) or ('ADV' in veh_id)):
            veh_labels.append(1)
        else:
            veh_labels.append(0)
            
    return veh_labels

def get_thresholding_performance(rec_error_dict,threshold=5.0):
    
    veh_ids = list(rec_error_dict.keys())
    
    true_labels = get_true_labels(veh_ids)
    
    max_rec_errors = get_max_rec_errors(rec_error_dict) #in order of veh_ids
    
    true_labels = np.array(true_labels)
    
    max_rec_errors = np.array(max_rec_errors)
    
    assigned_labels = max_rec_errors > threshold
    
    y = np.array(true_labels)
    
    pred = assigned_labels
        
    fpr, tpr, thresholds = metrics.roc_curve(y, pred, pos_label=1)
    
    return fpr, tpr, thresholds
    
def get_thresholding_AUC(rec_error_dict):
    
    veh_ids = list(rec_error_dict.keys())
    
    true_labels = get_true_labels(veh_ids)
    
    max_rec_errors = get_max_rec_errors(rec_error_dict) #in order of veh_ids
    
    true_labels = np.array(true_labels)
    
    max_rec_errors = np.array(max_rec_errors)
    
    fpr, tpr, thresholds = metrics.roc_curve(true_labels, max_rec_errors, pos_label=1)
    

    return metrics.auc(fpr, tpr)


def get_thresholding_F1(rec_error_dict,threshold=5.0):
    
    veh_ids = list(rec_error_dict.keys())
    
    true_labels = get_true_labels(veh_ids)
    
    max_rec_errors = get_max_rec_errors(rec_error_dict) #in order of veh_ids
    
    true_labels = np.array(true_labels)
    
    max_rec_errors = np.array(max_rec_errors)
    
    assigned_labels = max_rec_errors > threshold
    
    y_true = np.array(true_labels)
    
    y_pred = np.array(assigned_labels)
    
    f1 = f1_score(y_true, y_pred, pos_label=1)
    
    return f1
```
